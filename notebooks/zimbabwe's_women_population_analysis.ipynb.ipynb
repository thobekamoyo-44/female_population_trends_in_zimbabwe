Creating A Decade column

```python
print(women_in_zim["Year"].min())
print(women_in_zim["Year"].max())
```

```python
women_in_zim["Decade"] = (women_in_zim["Year"]// 10 * 10).astype(str) + "s"
```

Female Population

```python
female_pop = women_in_zim[women_in_zim["Indicator Name"] == "Population, female"]

female_pop_by_decade = female_pop.groupby("Decade")["Value"].mean()/1000000

female_pop_by_decade.round(2)
```

The average female population increased from 2.23 million in the 1960s to 8.50 million in the 2020s, nearly four times the 1960s level."

```python
female_pop_by_decade_plot = female_pop_by_decade.round(2)

female_pop_by_decade_plot.plot(kind="line", figsize= (6, 4), marker = "o")

plt.title("FEMALE POPULATION GROWTH IN ZIMBABWE (1960 to 2025)")
plt.xlabel("Decades")
plt.ylabel("Female Population (million)")
plt.tight_layout()
plt.show()
```

The Graph  above shows that Zimbabwe's female population grew continously and steady from an average of 2.23 million in the 1960s to 8.50 million in the 2020s , almost Quadruplating in six decades

Figuring out what led to  to this growth in female population

Fertility Rate

```python
TFR =  women_in_zim[women_in_zim["Indicator Name"].str.contains("total fertility rate", case=False, na=False )]
TFR_by_decade = TFR.groupby("Decade")["Value"].mean()
print(f"Fertility By Decade : \n{TFR_by_decade.round(2)}")
```

```python
TFR_by_decade_plot = TFR_by_decade.round(2)

TFR_by_decade_plot.plot(kind="line", figsize= (6, 3), marker = "o", color= "red")

plt.title("Total Fertility Rate in Zimbabwe by Decade")
plt.xlabel("Decade")
plt.ylabel("Fertility Rate (births per women)")
plt.tight_layout()
plt.show()
```

```python
TFR_by_year = TFR.groupby("Year")["Value"].mean()
print(f"Fertility By Year : \n{TFR_by_year.round(2)}")

TFR_by_year_plot = TFR_by_year.round(2)

TFR_by_year_plot.plot(kind="line", figsize= (6, 3), marker = "o", color= "red")

plt.title("Total Fertility Rate in Zimbabwe by Year")
plt.xlabel("Year")
plt.ylabel("Fertility Rate (births per women)")
plt.tight_layout()
plt.show()
```

The average total fertility rate declined from 4.25 births per woman in the 1990s to 3.94 in the 2000s, before increasing slightly to 4.24 in the 2010s.

```python
TFR_by_wealth =  TFR.groupby("Indicator Name")["Value"].mean()
print(TFR_by_wealth.round(2))
```

```python
TFR_by_wealth_plot = TFR_by_wealth.round(2)

TFR_by_wealth_plot.plot(kind= 'barh',figsize=(10,4), color="purple")

plt.title("Total Fertility Rate by Wealth Quintile")
plt.xlabel("Fertility Rate (births per women)") 
plt.ylabel("Wealth Quintile") 
plt.tight_layout() 
plt.show()                                      
```

The average fertility rate among women in the poorest quintile was more than twice that of women in the richest quintile

Factors affecting fertility rate in different wealth quantiles

Contraception

```python
contraception = women_in_zim[women_in_zim["Indicator Name"].str.contains("Current use of contraception (any method) (% of married women)", case=False, na=False, regex=False)]
contra_usage_by_decade = contraception.groupby("Decade")["Value"].mean()
print(contra_usage_by_decade.round(2))
```

```python
contra_usage_by_decade_plot = contra_usage_by_decade.round(2)

contra_usage_by_decade_plot.plot(
    kind="line",
    figsize=(8, 5),
    marker="o",
    color="pink"
)

plt.title("Contraceptive Use Among Married Women in Zimbabwe")
plt.xlabel("Decade")
plt.ylabel("Percentage of married women using Contraception")
plt.tight_layout()
plt.show()
```

```python
contra_usage_by_year = contraception.groupby("Year")["Value"].mean()
print(contra_usage_by_year.round(2))

contra_usage_by_year_plot = contra_usage_by_year.round(2)

contra_usage_by_year_plot.plot(
    kind="line",
    figsize=(8, 5),
    marker="o",
    color="pink"
)

plt.title("Contraceptive Use Among Married Women in Zimbabwe")
plt.xlabel("Year")
plt.ylabel("Percentage of married women using Contraception")
plt.tight_layout()
plt.show()
```

Contraception usage grew by 12.13 % from 1990  t0 2000, however theres only 1.54% increase from the 2000s to the 2010s
13.67 % more women are using contraceptives in 2010s than in the 1990s

```python
contra_usage_by_wealth = contraception.groupby("Indicator Name")["Value"].mean()
print(contra_usage_by_wealth.round(2))
```

Linking the tables 
fertility rate vs use of contraception (in different classes of wealth)

```python
fertility_clean = pd.Series(TFR_by_wealth.values, index=["Q1", "Q2", "Q3", "Q4", "Q5"])
contra_clean = pd.Series(contra_usage_by_wealth.values, index=["Q1", "Q2", "Q3", "Q4", "Q5"])
```

```python
fertility_vs_contraception = pd.DataFrame({"fertility Rate":fertility_clean,
                                           "Contraception Use":contra_clean})

print(fertility_vs_contraception.round(2))
```

```python
print(f"Correlation: \n{fertility_vs_contraception.corr().round(2)}")
```

correlation

```python
x = ["Q1\nPoorest", "Q2", "Q3", "Q4", "Q5\nRichest"]

fig, ax1 =plt.subplots(figsize=(9,5))

# fertility (Left axis in Pink)
ax1.plot(x, fertility_clean.values, linewidth=3, marker="o", color="pink")
ax1.set_ylabel("Fertility(births per woman)", color="pink")
ax1.set_ylim(0,6)
ax1.tick_params(axis="y", labelcolor="pink")

#Contraception (right axis in Orange)
ax2=ax1.twinx()
ax2.plot(x, contra_clean.values, linewidth =3, marker="s", color= "orange")
ax2.set_ylabel("Contraception Use (% married women)", color="orange")
ax2.set_ylim(0,80)
ax2.tick_params(axis= "y",labelcolor="orange")


plt.title("Wealth, Fertility and Contraception in Zimbabwe (1990s - 2010s)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    contra_clean.values,
    fertility_clean.values
)

for q, x, y in zip(
    ["Q1", "Q2", "Q3", "Q4", "Q5"],
    contra_clean.values,
    fertility_clean.values
):
    plt.annotate(q, (x, y))

plt.title("Contraceptive Use and Fertility by Wealth Quintile")
plt.xlabel("Contraceptive use (% of married women)")
plt.ylabel("Fertility rate (births per woman)")
plt.tight_layout()
plt.show()
```

Contraceptive Use and Fertility by Wealth Quintile graph shows a clear inverse relationship between wealth, contraception use  and  fertility. as we move from poorest quintile (Q1) to richest(Q5) ,the use of contraceptions increases while the total fertility declines.this suggest that greater access to and use of contraception is associated with lower fertility rate.

Access to health facilities(distance as the problem)

```python
health_facility_access = women_in_zim[women_in_zim["Indicator Name"].str.contains("Problems in accessing health care (distance to health facility) (% of women)", case= False, na=False, regex=False)]
health_facility_access_by_decade = health_facility_access.groupby("Decade")["Value"].mean()
print(health_facility_access_by_decade)
```

```python
health_facility_access_by_decade_plot = health_facility_access_by_decade.round(2)

health_facility_access_by_decade_plot.plot(
    kind="line",
    figsize=(8, 5),
    marker="o"
)

plt.title("Difficulty Accessing Healthcare Due to Distance in Zimbabwe")
plt.xlabel("Decade")
plt.ylabel("Women reporting difficulty (%)")
plt.tight_layout()
plt.show()
```

```python
health_facility_access_by_year = health_facility_access.groupby("Year")["Value"].mean()
print(health_facility_access_by_year)

health_facility_access_by_year_plot = health_facility_access_by_year.round(2)

health_facility_access_by_year_plot.plot(
    kind="line",
    figsize=(8, 5),
    marker="o"
)

plt.title("Difficulty Accessing Healthcare Due to Distance in Zimbabwe")
plt.xlabel("Year")
plt.ylabel("Women reporting difficulty (%)")
plt.tight_layout()
plt.show()
```

The average percentage of women reporting difficulty accessing healthcare facilities due to distance increased from 33.36% in the 1990s to 44.86% in the 2000s. This increase coincided with Zimbabwe's severe economic crisis, which contributed to the deterioration of healthcare services, including the closure of some public health facilities and shortages of healthcare workers. These circumstances may have contributed to increased difficulties accessing healthcare.

In the 2010s, the percentage declined to 36.86%. This decline coincided with economic stabilisation following the introduction of the multicurrency system in 2009 and efforts to restore healthcare services. However, the percentage remained higher than in the 1990s, suggesting that challenges with physical access to healthcare persisted.

```python
health_facility_access_by_wealth = health_facility_access.groupby("Indicator Name")["Value"].mean()

print(health_facility_access_by_wealth)
```

```python
health_facility_access_by_wealth.plot(
    kind="barh",
    figsize=(6, 5)
)

plt.title("Difficulty Accessing Health Care Due to Distance")
plt.xlabel("Percentage of women")
plt.ylabel("Wealth quintile")
plt.tight_layout()
plt.show()
```

Women in the poorest wealth quintile reported substantially greater difficulty accessing health care because of distance, while contraceptive use was lower and fertility was higher. This pattern suggests an association between healthcare access, contraceptive use and fertility.
